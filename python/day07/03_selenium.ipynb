```python
SSR vs CSR
Server Side Rendering : 웹 화면을 서버에서 만들어서 클라이언트에 보냄
-> request를 이용해서 화면 정보를 가져올 수 있음

Client Side Rendering : 웹 화면을 클라이언트에서 구성(리액트)
-> request를 이용해서 화면 정보를 못 가져옴
-> selenium을 이용해서 화면 정보를 가져올 수 있음
    - 브라우저를 띄움
    - 해당 사이트 접속
    - 다 로딩될때까지 기다림(지연시간)
    - 화면 정보를 가져옴
```

```python
from selenium import webdriver
import time

# 브라우저가 바로 닫히는 것을 방지하는 옵션
options = webdriver.ChromeOptions()
options.add_experimental_option("detach", True)

# 브라우저 실행
driver = webdriver.Chrome(options=options)


url = "https://www.naver.com"
# 페이지 이동
driver.get(url)

# 페이지 로딩을 위해 3초 대기 : 충분한 로딩시간
time.sleep(5)

# html 코드를 읽어옴
html_source = driver.page_source

# 요소 탐색 및 동작 가능
# 예시 생략


with open('naver.html', 'wt', encoding='utf-8') as f:
    f.write(html_source)
    
# 모든 작업 완료 후 브라우저 종료
driver.quit()
```
