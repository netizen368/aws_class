```python
# Phone 클래스를 추가
# 모델명(model), 가격(price)
# 기능 : 생성자 - 모델명과 가격을 초기화, 모델조회(display)  모델과 가격
class Phone:

    # 생성자
    def __init__(self, model : str, price : int):
        self.__model : str = model
        self.__price : int = price

    # 모델 조회
    def display(self):
        print(f"모델명 : {self.__model}")
        print(f"가격 : {self.__price:,}원")

# IPhone 클래스를 추가(Phone 클래스를 상속)
# 페이스아이디적용여부(face_id)
# 기능 : 생성자 - 모델명과 가격, 페이스아이디 적용 여부를 초기화,
#        모델조회(display) - 모델명, 가격, 페이스적용여부
class IPhone(Phone):
    # 생성자
    def __init__(self, model : str, price : int, face_id : bool):
        super().__init__(model, price)
        self.__face_id : bool = face_id

    def display(self):
        line_num = 20
        print("-" * line_num)
        super().display()
        print(f"얼굴 인식 지원 : {"0" if self.__face_id else "X"}")
        print("-" * line_num)
        
# Galaxy 클래스를 추가(Phone 클래스를 상속)
# 펜 지원(s_pan)
# 기능 : 생성자 - 모델명, 가격, 펜지원여부를 초기화
#        모델조회(display) - 모델명, 가격, 펜지원 여부
class Galaxy(Phone):
    def __init__(self, model : str, price : int, s_pan : bool):
        super().__init__(model, price)
        self.__s_pan : bool = s_pan

    def display(self):
        line_num = 20
        print("-" * line_num) # 데코
        super().display()
        print(f"펜지원 여부 : {"0" if self.__s_pan else "X"}")
        print("-" * line_num) # 데코
        
# IPhone : iPhone15, 1250000, True
# IPhone : iPhoneSE, 650000, False
# Galaxy : S24 Ultra, 1690000, True
# Galaxy : Z Flip 5, 1390000, False
phones = [
    IPhone("iPhone15", 1250000, True),
    IPhone("iPhoneSE", 650000, False),
    IPhone("S24 Ultra", 1690000, True),
    IPhone("Z Flip 5", 1390000, False)
]
# 각 핸드폰을 조회
for phone in phones:
    phone.display()
```
